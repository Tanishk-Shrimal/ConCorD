```python
from concord_lqcd.core import (
    Correlator,
    Operator,
    c, cbar,
    u, ubar,
    d, dbar,
    gamma5,
    gammaX,
    gammaZ,
    reset_registry,
)

reset_registry()
```

```python
# Pseudo-scalar Pion operator: \bar{d}(x) \gamma_5 u(x) at momentum p = (0, 0, 0)
pion = Operator(dbar("x", "a", "alpha") * gamma5("alpha", "beta") * u("x", "a", "beta"), momentum=(0, 0, 0))

# Vector Rho operator: \bar{d}(x) \gamma_x u(x) at momentum p = (0, 0, 0)
rho = Operator(dbar("x", "a", "alpha") * gammaX("alpha", "beta") * u("x", "a", "beta"), momentum=(0, 0, 0))

pion_dagger = pion.dagger()

print("Pion:", pion)
print("Rho:", rho)
print("Pion Dagger:", pion_dagger)
```

```python
corr = Correlator(pion, pion_dagger)
```

```python
from concord_lqcd.backend.runtime import (
    RuntimeCompilationConfig,
    compile_correlators,
)
Nv = 64   # whatever your ensemble uses

compilation = compile_correlators(
    {"pion": corr},
    RuntimeCompilationConfig(
        mode_size=Nv,
        scalar_type="complex128",
        element_bytes=16,
        contraction_strategy="greedy",
        schedule_strategy="memory_aware",
        reuse_strategy="best_fit",
    ),
)
```

```python
for resource_id, recipe in compilation.required_propagation_resources.items():
    print()
    print("resource:", resource_id)
    print("kind:", recipe.resource_kind)
    print("flavors:", recipe.segment_flavors)
    print("source representation:", recipe.source_representation)
    print("sink representation:", recipe.sink_representation)
    print("generalized:", recipe.is_generalized)
    print("n insertions:", len(recipe.insertion_specs))
```
